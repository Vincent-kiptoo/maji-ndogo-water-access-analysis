## Discrepancy records reprenting unsafe but reported as safe

```python
%%sql
SELECT *
FROM discrepancy
LIMIT 5
```

```python
%%sql
-- This query dertimines if sources with failure type of Broken taps and long queues were affected by records of the officials who had discrepancies
SELECT
    source_id,
    employee_score,
    auditor_score,
    score_difference
FROM discrepancy
WHERE failure_type IN ('Broken Tap', 'Long Queue (>30)')
```

```python
%%sql
-- determines the total population AND water sources affected by the officials with discrepancy
SELECT
    location_id,
    province_name,
    town_name,
    failure_type,
    SUM(population_affected) AS total_population_affected,
    RANK() OVER (PARTITION BY province_name ORDER BY SUM(population_affected) DESC) AS `rank`
FROM
    safe_but_unsafe_water
GROUP BY 
    location_id,
    province_name,
    town_name,
    failure_type;
```
